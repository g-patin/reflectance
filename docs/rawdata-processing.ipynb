In this section, we will show you how to process rawdata files into and convert them into standard *interim* files, that are required to use the `reflectance` package.

The current version of the `reflectance` package can only process raw files obtained with the following devices:

- Avantes, 
- Tidas, S400 and S800 spectrometers
- Malvern Panalytica, ASD FieldSpec 4 Hi-Res spectrometer

Don't hesitate to contact us to discuss the possibility to process raw files obtained with other microfading systems.

Processing the rawdata files consists of two main steps:

    1. Get a list of your raw files

    2. Run the process_rawdata() function

# Run processing function

## Basic use

The code below describes the most basic way to process raw files. It does not make use of the information stored in the configuration file and database files, which is why the parameter *db* should be set to False. 

To run the function, you will only need to give a list of your filepaths and the name of your device. The latter will have to be one of the acronyms below, depending on the files you are processing. You cannot process at the same time, rawdata files that have been obtained from different devices.

List of device acronyms:

- 'ASD' : Malvern Panalytical spectrometer, ASD LabSpec 4
- 'AVT' : Avantes spectrometer, AvaSpec-ULS-EVO-RS, 2201504U1
- 'Tidas' : Tidas spectrometer, S400 or S800

This way of processing the raw files is not recommended. While it does not impact the processing of the numerical data (colorimetric and spectral values), the metadata are almost non-existent. The real values of our analyses lies in our abilities to connect them with tangible aspects (objects, persons, devices), numerical values without context are meaningless. 


```python
import reflectance as rf
```

```python
# For this tutorial, I am using raw files available from the datasets of the reflectance package
files_rs = rf.get_datasets(rawfiles=True)

# if db is set to True, make sure that you have registered the device
rf.process_rawdata(files=files_rs, device='AVT', db=False)
```
