This section will show you how to create an instance of the Reflectance Spectroscopy (`RS`) class. The creation of the `RS` class is a two-steps procedure as described below:

# 1. Get the interim files

```python
import reflectance as rf
```

```python
# Here we are using the test datasets, but ideally you should use your own interim files.
files_rs = rf.get_datasets()
```

# 2. Create the `RS` class object

```python
rs = rf.RS(files=files_rs)
rs
```
