# Configuration steps

1. Configure the 

There are three different verbs you can use and that define the actions you can perform:

- **get** : retrieve information about the configuration
- **set** : configure the parameters registered in the config_info.json file
- **remove**: delete the value of a parameter that you registered

# Get functions

The *get* functions (described in Table 1) enables you to retrieve information about the configuration. The main get function is `get_config()` which will return all the parameters you have saved. The content of the file consists of seven dictionaries (databases, devices, colorimetry, comments, filenaming, institution, systems) that you can used as parameter inside the `get_config()` function. 

For example, to retrieve the information about the devices only, you would type:

```python
rf.get_config('devices')
```

At the beginning, when you haven't configured any parameters, all the dictionaries are empty. You will populate them with the *set* functions (see below).


Table 1. Description of the *get* functions.

|`get config()` | returns the content of the config_info.json file
|`get_institution_info()` | returns information about your institution
